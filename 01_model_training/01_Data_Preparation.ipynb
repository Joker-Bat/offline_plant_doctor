```python
# This cell downloads and prepares the dataset from Kaggle

# Define the dataset path on Kaggle and the local directory
dataset_name = "vipoooool/new-plant-diseases-dataset"
data_dir = "data"

# Create the target directory if it doesn't exist
os.makedirs(data_dir, exist_ok=True)

# Download the dataset using the Kaggle CLI
# The -p flag specifies the path, and --unzip automatically extracts it.
print(f"Downloading dataset '{dataset_name}' from Kaggle...")
!kaggle datasets download -d {dataset_name} -p {data_dir} --unzip

print("\nDownload and extraction complete.")

# --- Verify the contents ---
print("\nVerifying the dataset structure:")
# The dataset contains two main folders, we're interested in the original images
# Note: The folder name in the archive has spaces.
unzipped_folder_name = os.path.join(data_dir, 'New Plant Diseases Dataset(Original)')

if os.path.exists(unzipped_folder_name):
     print(f"Found dataset in: {unzipped_folder_name}")
     print("Listing a few disease categories:")
     # List the first 5 directories (plant disease categories)
     for item in os.listdir(unzipped_folder_name)[:5]:
         print(f"- {item}")
else:
    # Let's check for other common names, sometimes they change
    alt_folder_name = os.path.join(data_dir, 'new-plant-diseases-dataset')
    if os.path.exists(alt_folder_name):
        print(f"Found dataset in: {alt_folder_name}")
        print("Listing a few disease categories:")
        for item in os.listdir(alt_folder_name)[:5]:
            print(f"- {item}")
    else:
        print("Error: Could not find the expected unzipped folder.")
        print("Listing contents of data directory for debugging:")
        !ls -l {data_dir}
```

```python
# Import the os library for handling file paths
import os

# Define the base directory using the name from our previous log
data_dir = "data"
base_dataset_dir = os.path.join(data_dir, 'New Plant Diseases Dataset(Augmented)')

# --- Check if the directory exists ---
if os.path.exists(base_dataset_dir):
    print(f"Successfully located the base dataset directory:\n{base_dataset_dir}\n")

    # Define the paths for our training and validation sets
    train_dir = os.path.join(base_dataset_dir, 'train')
    validation_dir = os.path.join(base_dataset_dir, 'valid')

    # --- Verify the existence of train and validation folders ---
    if os.path.exists(train_dir) and os.path.exists(validation_dir):
        print("Found 'train' and 'valid' directories inside.")

        # Get the list of disease classes from the subdirectories in the training folder
        classes = sorted(os.listdir(train_dir))
        num_classes = len(classes)
        print(f"\nFound {num_classes} disease categories (classes).")
        print("A few examples:", classes[:5])

        # --- Count the number of images ---
        # We use a generator expression for efficiency - it avoids loading all filenames into memory at once
        total_train_images = sum(len(files) for _, _, files in os.walk(train_dir))
        total_validation_images = sum(len(files) for _, _, files in os.walk(validation_dir))

        print(f"\nTotal number of training images: {total_train_images}")
        print(f"Total number of validation images: {total_validation_images}")
        print("\n-------------------------------------------")
        print("✅ Data preparation and verification complete!")
        print("-------------------------------------------")

    else:
        print("\n❌ Error: Could not find 'train' and/or 'valid' directories inside the base folder.")
        print("Let's see what's actually inside:")
        !ls -l "{base_dataset_dir}"

else:
    print(f"❌ Error: The base directory '{base_dataset_dir}' still cannot be found.")
```
